## Question 1

Viết `describe_values(values)` trả về dictionary gồm population mean và population standard deviation với hai key `mean`, `std`.

```python
def describe_values(values):
    # Demo implementation. Replace this body with the student starter code.
    array = np.asarray(values, dtype=float)
    return {
        "mean": float(array.mean()),
        "std": float(array.std(ddof=0)),
    }
```

## Question 2

Viết `revenue_by_product(records)` nhận một list các record và trả về tổng revenue theo product dưới dạng dictionary.

```python
def revenue_by_product(records):
    # Demo implementation. Replace this body with the student starter code.
    if not records:
        return {}

    frame = pd.DataFrame.from_records(records)
    totals = frame.groupby("product", sort=True)["revenue"].sum()
    return {name: float(value) for name, value in totals.items()}
```
